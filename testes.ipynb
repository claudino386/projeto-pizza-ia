```python
import pandas as pd

df = pd.read_csv("pizzas.csv")

df
```

```python
df.columns = df.columns.str.strip()

df.plot(kind="scatter", x="diametro", y="preco");
```

```python
from sklearn.linear_model import LinearRegression

# Definindo X (entrada/diâmetro) e y (saída/preço)
X = df[['diametro']]
y = df['preco']

# Criando e treinando a IA
modelo = LinearRegression()
modelo.fit(X, y)

print("Inteligência Artificial treinada com sucesso!")
```

```python
# Coloque o número do diâmetro (ex: 50) dentro dos colchetes
dados_teste = pd.DataFrame({'diametro': [50]})
preco_previsto = modelo.predict(dados_teste)

print(f"Preço estimado pela IA: R$ {preco_previsto[0]:.2f}")
```
